**Mục tiêu bài giảng:**

3. Thực hành xây dựng 3 bài toán kinh điển: Hồi quy Tuyến tính (Linear Regression), Phân lớp (Logistic Regression), Phân cụm (K-Means Clustering).

---
### 🎯 Kiến thức cốt lõi dành cho học viên:
- **Supervised Learning**: Dữ liệu có nhãn ($X \to y$). Gồm Regression ($y$ liên tục) và Classification ($y$ rời rạc).
- **Unsupervised Learning**: Dữ liệu không có nhãn ($X$). Mục tiêu tìm cấu trúc ẩn (Clustering, Dimensionality Reduction).
- **Evaluation**: RMSE, $R^2$ cho Regression; Accuracy, Confusion Matrix cho Classification.


## 1. Import các Thư viện Tiêu chuẩn trong Data Science
Chúng ta sẽ import các thư viện cốt lõi:
- `numpy`: Thao tác mảng số học đa chiều.
- `pandas`: Xử lý bảng dữ liệu (DataFrame).
- `matplotlib.pyplot` & `seaborn`: Trực quan hóa dữ liệu và kết quả.
- `sklearn`: Thư viện Machine Learning chuẩn mực hàng đầu hiện nay.


````python
# Import các thư viện cần thiết
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình font và style biểu đồ trực quan
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)

# Import các module từ Scikit-Learn
from sklearn.datasets import make_regression, make_classification, make_blobs
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.cluster import KMeans
from sklearn.metrics import r2_score, root_mean_squared_error, accuracy_score, confusion_matrix

print('✅ Đã nạp thành công toàn bộ thư viện cần thiết!')
````

📌 **Expected Output:**
```
✅ Đã nạp thành công toàn bộ thư viện cần thiết!
```


## 2. Phần 1: Học Có Giám Sát - Bài Toán Hồi Quy (Supervised Regression)
Chúng ta tạo một tập dữ liệu nhân tạo $y = w_1 x + w_0 + \epsilon$ có độ nhiễu và huấn luyện mô hình Linear Regression.


````python
# 1. Sinh dữ liệu hồi quy nhân tạo (100 mẫu, 1 đặc trưng)
X_reg, y_reg = make_regression(n_samples=100, n_features=1, noise=15.0, random_state=42)

# 2. Phân chia Train/Test theo tỷ lệ 80/20
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# 3. Khởi tạo và huấn luyện mô hình Linear Regression
reg_model = LinearRegression()
reg_model.fit(X_reg_train, y_reg_train)

# 4. Dự đoán và Đánh giá
y_reg_pred = reg_model.predict(X_reg_test)
rmse = root_mean_squared_error(y_reg_test, y_reg_pred)
r2 = r2_score(y_reg_test, y_reg_pred)

print(f'Hệ số góc w1:      {reg_model.coef_[0]:.4f}')
print(f'Hệ số chặn w0:     {reg_model.intercept_:.4f}')
print(f'RMSE trên Test set: {rmse:.4f}')
print(f'R² Score:          {r2:.4f}')
````

📌 **Expected Output:**
```
Hệ số góc w1:      45.4957
Hệ số chặn w0:     0.1488
RMSE trên Test set: 15.3119
R² Score:          0.8742
```


## 3. Phần 2: Học Có Giám Sát - Bài Toán Phân Lớp (Supervised Classification)
Xây dựng mô hình phân lớp nhị phân (Binary Classification) để phân biệt 2 nhóm đối tượng.


````python
# 1. Sinh dữ liệu phân lớp 2 nhóm (200 mẫu, 2 đặc trưng)
X_clf, y_clf = make_classification(n_samples=200, n_features=2, n_redundant=0, n_informative=2, n_clusters_per_class=1, random_state=42)

# 2. Phân chia Train/Test (80/20)
X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42)

# 3. Chuẩn hóa đặc trưng bằng StandardScaler
scaler = StandardScaler()
X_clf_train_scaled = scaler.fit_transform(X_clf_train)
X_clf_test_scaled = scaler.transform(X_clf_test)

# 4. Huấn luyện Logistic Regression
clf_model = LogisticRegression(random_state=42)
clf_model.fit(X_clf_train_scaled, y_clf_train)
y_clf_pred = clf_model.predict(X_clf_test_scaled)

# 5. Đánh giá
acc = accuracy_score(y_clf_test, y_clf_pred)
cm = confusion_matrix(y_clf_test, y_clf_pred)
cv_scores = cross_val_score(clf_model, scaler.transform(X_clf), y_clf, cv=5)

print(f'Độ chính xác (Accuracy): {acc * 100:.2f}%')
print(f'5-Fold Cross-Val Acc:    {cv_scores.mean() * 100:.2f}% (+/- {cv_scores.std()*100:.2f}%)')
print('Confusion Matrix:')
print(cm)
````

📌 **Expected Output:**
```
Độ chính xác (Accuracy): 87.50%
5-Fold Cross-Val Acc:    83.50% (+/- 6.04%)
Confusion Matrix:
[[18  5]
 [ 0 17]]
```


## 4. Phần 3: Học Không Giám Sát - Phân Cụm Dữ Liệu (K-Means Clustering)
Khám phá các cụm dữ liệu tự nhiên mà không cần nhãn mục tiêu ($y$).


````python
# 1. Sinh dữ liệu gồm 3 cụm tự nhiên
X_cluster, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)

# 2. Huấn luyện K-Means với k=3 cụm
kmeans = KMeans(n_clusters=3, n_init='auto', random_state=42)
cluster_labels = kmeans.fit_predict(X_cluster)

print('Tọa độ 3 tâm cụm (Cluster Centers):')
print(kmeans.cluster_centers_)
print(f'Inertia (Tổng bình phương khoảng cách tới tâm): {kmeans.inertia_:.2f}')
````

📌 **Expected Output:**
```
Tọa độ 3 tâm cụm (Cluster Centers):
[[-2.60842567  9.03771305]
 [-6.88302287 -6.96320924]
 [ 4.72565847  2.00310936]]
Inertia (Tổng bình phương khoảng cách tới tâm): 362.79
```


## 5. Phần 4: Trực Quan Hóa Tổng Hợp Cả 3 Bài Toán
Vẽ biểu đồ kết quả trực quan trên 3 đồ thị con (Subplots) để học viên đối chiếu trực tiếp.


````python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Đồ thị 1: Hồi quy Tuyến tính
axes[0].scatter(X_reg_train, y_reg_train, color='blue', alpha=0.6, label='Train')
axes[0].scatter(X_reg_test, y_reg_test, color='green', alpha=0.6, label='Test')
x_line = np.linspace(X_reg.min(), X_reg.max(), 100).reshape(-1, 1)
axes[0].plot(x_line, reg_model.predict(x_line), color='red', linewidth=2, label=f'Fit (R²={r2:.2f})')
axes[0].set_title('1. Supervised: Linear Regression', fontweight='bold')
axes[0].legend()

# Đồ thị 2: Phân lớp
axes[1].scatter(X_clf_test_scaled[y_clf_test==0, 0], X_clf_test_scaled[y_clf_test==0, 1], color='red', label='Class 0')
axes[1].scatter(X_clf_test_scaled[y_clf_test==1, 0], X_clf_test_scaled[y_clf_test==1, 1], color='blue', label='Class 1')
axes[1].set_title(f'2. Supervised: Logistic Regression (Acc={acc*100:.1f}%)', fontweight='bold')
axes[1].legend()

# Đồ thị 3: Phân cụm K-Means
axes[2].scatter(X_cluster[:, 0], X_cluster[:, 1], c=cluster_labels, cmap='viridis', alpha=0.7)
axes[2].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroids')
axes[2].set_title('3. Unsupervised: K-Means (k=3)', fontweight='bold')
axes[2].legend()

plt.tight_layout()
plt.show()
````

📌 **Expected Output:**
- Hiển thị đồng thời 3 đồ thị: Đường thẳng hồi quy màu đỏ khớp dữ liệu; Vùng phân lớp đỏ/xanh; 3 cụm màu kèm 3 dấu X đỏ tại các tâm cụm.
