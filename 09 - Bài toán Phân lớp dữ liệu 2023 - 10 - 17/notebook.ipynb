# 📘 Bài Giảng Thực Hành: Phân Lớp Dữ Liệu & Các Độ Đo Đánh Giá (Classification Metrics)
**Mục tiêu bài giảng:**
1. Hiểu toàn diện cấu trúc của Ma trận Nhầm lẫn (Confusion Matrix: TP, TN, FP, FN).
2. Phân tích ý nghĩa và ứng dụng thực tế của: Accuracy, Precision, Recall, Specificity, F1-Score, ROC-AUC.
3. Sử dụng các công cụ hiển thị chuẩn Scikit-Learn mới: `ConfusionMatrixDisplay`, `RocCurveDisplay`, `PrecisionRecallDisplay`.

---
### 📊 Tập dữ liệu thực hành: `Social_Network_Ads.csv`


## 1. Nạp Thư viện và Chuẩn Bị Dữ Liệu


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, fbeta_score,
    roc_auc_score, classification_report, ConfusionMatrixDisplay, RocCurveDisplay, PrecisionRecallDisplay
)

dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, [2, 3]].values  # Age, EstimatedSalary
y = dataset.iloc[:, 4].values       # Purchased (0 hoặc 1)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)
dataset.head()
````

📌 **Expected Output:**
Bảng dữ liệu gồm các cột User ID, Gender, Age, EstimatedSalary, Purchased.


## 2. Huấn luyện Mô hình & Tính Toán Các Độ Đo Phân Lớp


````python
clf = LogisticRegression(random_state=0).fit(X_train_scaled, y_train)
y_pred = clf.predict(X_test_scaled)
y_prob = clf.predict_proba(X_test_scaled)[:, 1]

cm = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cm.ravel()

print(f'True Positives (TP):   {tp} | True Negatives (TN):  {tn}')
print(f'False Positives (FP):  {fp} | False Negatives (FN): {fn}')
print('-'*45)
print(f'Accuracy:    {accuracy_score(y_test, y_pred)*100:.2f}%')
print(f'Precision:   {precision_score(y_test, y_pred)*100:.2f}%')
print(f'Recall:      {recall_score(y_test, y_pred)*100:.2f}%')
print(f'F1-Score:    {f1_score(y_test, y_pred):.4f}')
print(f'ROC-AUC:     {roc_auc_score(y_test, y_prob):.4f}')
````

📌 **Expected Output:**
```

---------------------------------------------

```


## 3. Trực quan hóa Bộ Đồ Thị Đánh Giá


````python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Confusion Matrix
ConfusionMatrixDisplay.from_estimator(clf, X_test_scaled, y_test, display_labels=['Not Buy', 'Buy'], cmap='Blues', ax=axes[0])
axes[0].set_title('Confusion Matrix', fontweight='bold')

# ROC Curve
RocCurveDisplay.from_estimator(clf, X_test_scaled, y_test, name='Logistic Reg', ax=axes[1])
axes[1].plot([0, 1], [0, 1], 'k--')
axes[1].set_title('ROC Curve (AUC = 0.954)', fontweight='bold')

# PR Curve
PrecisionRecallDisplay.from_estimator(clf, X_test_scaled, y_test, name='Logistic Reg', ax=axes[2])
axes[2].set_title('Precision-Recall Curve', fontweight='bold')

plt.tight_layout()
plt.show()
````

📌 **Expected Output:**
Hiển thị 3 biểu đồ trực quan: Ma trận nhầm lẫn xanh lam, Đường cong ROC với AUC=0.954 và Đường cong Precision-Recall.
