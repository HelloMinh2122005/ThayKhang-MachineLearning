# 📘 Bài Giảng Thực Hành: Cây Quyết Định Cho Phân Lớp (Decision Tree Classification)
**Mục tiêu bài giảng:**
1. Hiểu các tiêu chuẩn phân nhánh: Entropy / Information Gain vs Gini Impurity.
2. Kiểm soát Overfitting bằng tham số `max_depth`.
3. Trực quan hóa cấu trúc cây toàn diện bằng `plot_tree` và xuất bộ luật `export_text`.

---
### 📊 Tập dữ liệu: `Social_Network_Ads.csv`


## 1. Nạp Dữ Liệu & Huấn Luyện Decision Tree Classifier


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.metrics import accuracy_score

dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, [2, 3]].values
y = dataset.iloc[:, 4].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)

# max_depth=3 để hạn chế cây mọc quá sâu gây Overfitting
classifier = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=0)
classifier.fit(X_train_scaled, y_train)
print(f'Accuracy trên Test Set: {accuracy_score(y_test, classifier.predict(X_test_scaled))*100:.2f}%')
```

📌 **Expected Output:**
`Accuracy trên Test Set: 94.00%`


## 2. Xuất Bộ Luật Cây (IF-THEN Rules) và Sơ Đồ Cây (plot_tree)


```python
print('=== BỘ LUẬT PHÂN NHÁNH CỦA CÂY QUYẾT ĐỊNH ===')
print(export_text(classifier, feature_names=['Age', 'Salary']))

plt.figure(figsize=(14, 8))
plot_tree(classifier, feature_names=['Age', 'Salary'], class_names=['Not Buy', 'Buy'], filled=True, rounded=True)
plt.title('Sơ đồ Cây Quyết Định (Decision Tree Diagram)', fontweight='bold')
plt.show()
```

📌 **Expected Output:**
Sơ đồ cây với các ô màu tô theo độ thuần nhất (Entropy) và số lượng mẫu tại từng nút lá.
