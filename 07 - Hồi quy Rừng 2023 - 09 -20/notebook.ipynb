# 📘 Bài Giảng Thực Hành: Hồi Quy Rừng Ngẫu Nhiên (Random Forest Regression)
**Mục tiêu bài giảng:**
1. Hiểu kỹ thuật Ensemble Learning: Bagging (Bootstrap Aggregating) và Lấy mẫu đặc trưng ngẫu nhiên.
2. Quan sát sự làm mịn của đường dự đoán khi tăng số lượng cây ($n\_estimators = 10, 50, 100, 300$).
3. Tối ưu tốc độ huấn luyện với tham số `n_jobs=-1`.

---
### 📊 Tập dữ liệu: `Position_Salaries.csv`


## 1. Nạp Thư viện và Dữ liệu


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:2].values
y = dataset.iloc[:, 2].values
dataset.head()
````

## 2. Huấn luyện Random Forest và Khảo Sát Số Lượng Cây


````python
n_trees_list = [10, 50, 100, 300]
models = {}
for n in n_trees_list:
    rf = RandomForestRegressor(n_estimators=n, criterion='squared_error', random_state=0, n_jobs=-1)
    rf.fit(X, y)
    pred_6_5 = rf.predict([[6.5]])[0]
    r2 = r2_score(y, rf.predict(X))
    models[n] = rf
    print(f'Số cây: {n:3d} | Dự đoán Level 6.5: ${pred_6_5:,.2f} | R² Score: {r2:.4f}')
````

📌 **Expected Output:**
```
Số cây:  10 | Dự đoán Level 6.5: $167,000.00 | R² Score: 0.9419
Số cây:  50 | Dự đoán Level 6.5: $161,600.00 | R² Score: 0.9634
Số cây: 100 | Dự đoán Level 6.5: $158,300.00 | R² Score: 0.9704
Số cây: 300 | Dự đoán Level 6.5: $160,333.33 | R² Score: 0.9669
```


## 3. Trực quan hóa So sánh Hiệu Ứng Bậc Thang Của Rừng


````python
X_grid = np.arange(min(X)[0], max(X)[0], 0.01).reshape(-1, 1)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, n in enumerate(n_trees_list):
    ax = axes[idx]
    rf_model = models[n]
    ax.scatter(X, y, color='red', s=60, label='Data')
    ax.plot(X_grid, rf_model.predict(X_grid), color='blue', label=f'{n} Trees')
    ax.set_title(f'Random Forest ({n} Trees)', fontweight='bold')
    ax.set_xlabel('Position Level')
    ax.set_ylabel('Salary ($)')
    ax.legend()

plt.tight_layout()
plt.show()
````

📌 **Expected Output:**
4 đồ thị con cho thấy khi tăng số lượng cây, các bậc thang dự đoán trở nên mịn và tự nhiên hơn nhiều.
