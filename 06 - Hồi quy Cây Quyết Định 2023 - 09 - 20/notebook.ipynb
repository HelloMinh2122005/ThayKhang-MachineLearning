# 📘 Bài Giảng Thực Hành: Hồi Quy Cây Quyết Định (Decision Tree Regression)
**Mục tiêu bài giảng:**
1. Hiểu bản chất phân đoạn không gian và hàm bậc thang (Step-Wise Constant Function) của Cây Quyết Định.
2. Cập nhật Scikit-Learn: Sử dụng `criterion='squared_error'` thay cho `criterion='mse'` đã bị loại bỏ.
3. Trực quan hóa cấu trúc cây bằng `plot_tree` và xuất bộ luật bằng `export_text`.

---
### 📊 Tập dữ liệu: `Position_Salaries.csv`


## 1. Nạp Thư viện và Dữ liệu


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree, export_text
from sklearn.metrics import r2_score

dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:2].values
y = dataset.iloc[:, 2].values
dataset.head()
````

## 2. Huấn luyện DecisionTreeRegressor


````python
# Lưu ý: criterion='squared_error' là chuẩn scikit-learn mới nhất
regressor = DecisionTreeRegressor(criterion='squared_error', random_state=0)
regressor.fit(X, y)

pred_6_5 = regressor.predict([[6.5]])[0]
print(f'Dự đoán mức lương Level 6.5: ${pred_6_5:,.2f}')
print('Giải thích: Do Level 6.5 thuộc khoảng [5.5, 6.5], cây trả về mức lương của Level 6 ($150,000).')
````

📌 **Expected Output:**
```
Dự đoán mức lương Level 6.5: $150,000.00
```


## 3. Cấu Trúc Nhánh Cây Quyết Định (Tree Rules)


````python
tree_text = export_text(regressor, feature_names=['Position_Level'])
print(tree_text)
````

📌 **Expected Output:**
Bộ luật phân nhánh IF-THEN dạng text hiển thị các điểm cắt ngưỡng phân nhánh.


## 4. Trực quan hóa Hàm Bậc Thang và Sơ Đồ Cây


````python
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Đồ thị 1: Hàm bậc thang trên lưới mịn 0.01
X_grid = np.arange(min(X)[0], max(X)[0], 0.01).reshape(-1, 1)
axes[0].scatter(X, y, color='red', s=80, label='Dữ liệu Thực tế')
axes[0].plot(X_grid, regressor.predict(X_grid), color='blue', linewidth=2, label='Đường Hồi quy Cây Quyết Định')
axes[0].scatter([6.5], [pred_6_5], color='purple', s=160, marker='*', zorder=6, label=f'Dự đoán 6.5 (${pred_6_5:,.0f})')
axes[0].set_title('Decision Tree Regression (Step-Wise Constant)', fontweight='bold')
axes[0].set_xlabel('Position Level')
axes[0].set_ylabel('Salary ($)')
axes[0].legend()

# Đồ thị 2: Sơ đồ cây
plot_tree(regressor, ax=axes[1], feature_names=['Level'], filled=True, rounded=True)
axes[1].set_title('Sơ đồ Cây Quyết Định (plot_tree)', fontweight='bold')

plt.tight_layout()
plt.show()
````

📌 **Expected Output:**
Hiển thị biểu đồ hàm bậc thang với các đoạn thẳng ngang nối tiếp và sơ đồ cây phân nhánh đầy đủ.
