# 📘 Bài Giảng Thực Hành: Hồi Quy Đa Thức (Polynomial Regression)
**Mục tiêu bài giảng:**
1. Hiểu cách mô hình hóa các mối quan hệ phi tuyến dạng đường cong.
2. Nắm vững cách sử dụng `PolynomialFeatures` kết hợp `LinearRegression` qua `Pipeline`.
3. Nhận biết hiện tượng Underfitting (bậc quá thấp) vs Overfitting (bậc quá cao).
4. Dự đoán giá trị tại điểm kiểm thử Level = 6.5 và trực quan hóa đường cong mượt.

---
### 📊 Tập dữ liệu thực hành: `Position_Salaries.csv`
- **Level (1-10)**: Cấp bậc vị trí từ Business Analyst (1) đến CEO (10).
- **Salary**: Mức lương tương ứng theo cấp bậc ($45k đến $1,000k).


## 1. Nạp Dữ liệu & Khởi tạo


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score

dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:2].values  # Level
y = dataset.iloc[:, 2].values    # Salary
dataset
````

📌 **Expected Output:**
```
            Position  Level   Salary
0   Business Analyst      1    45000
1   Junior Consultant      2    50000
2   Senior Consultant      3    60000
3            Manager      4    80000
4    Country Manager      5   110000
5     Region Manager      6   150000
6            Partner      7   200000
7     Senior Partner      8   300000
8            C-level      9   500000
9                CEO     10  1000000
```


## 2. Huấn luyện Mô hình Tuyến tính vs Hồi quy Đa thức (Bậc 2, 3, 4)


````python
# Baseline: Tuyến tính đơn
lin_reg = LinearRegression().fit(X, y)

# Đa thức bậc 2, 3, 4 bằng make_pipeline
poly_reg_2 = make_pipeline(PolynomialFeatures(degree=2), LinearRegression()).fit(X, y)
poly_reg_3 = make_pipeline(PolynomialFeatures(degree=3), LinearRegression()).fit(X, y)
poly_reg_4 = make_pipeline(PolynomialFeatures(degree=4), LinearRegression()).fit(X, y)

# Dự đoán cho ứng viên Level = 6.5
test_level = np.array([[6.5]])
print(f'Linear Regression (Bậc 1): ${lin_reg.predict(test_level)[0]:,.2f} | R² = {r2_score(y, lin_reg.predict(X)):.4f}')
print(f'Polynomial (Bậc 2):        ${poly_reg_2.predict(test_level)[0]:,.2f} | R² = {r2_score(y, poly_reg_2.predict(X)):.4f}')
print(f'Polynomial (Bậc 3):        ${poly_reg_3.predict(test_level)[0]:,.2f} | R² = {r2_score(y, poly_reg_3.predict(X)):.4f}')
print(f'Polynomial (Bậc 4):        ${poly_reg_4.predict(test_level)[0]:,.2f} | R² = {r2_score(y, poly_reg_4.predict(X)):.4f}')
````

📌 **Expected Output:**
```
Linear Regression (Bậc 1): $330,378.79 | R² = 0.6690
Polynomial (Bậc 2):        $189,498.11 | R² = 0.9162
Polynomial (Bậc 3):        $133,259.47 | R² = 0.9812
Polynomial (Bậc 4):        $158,862.47 | R² = 0.9974
```


## 3. Trực quan hóa Đường Cong Đa Thức Độ Phân Giải Cao


````python
# Tạo lưới điểm mịn từ 1 đến 10
X_grid = np.arange(min(X)[0], max(X)[0], 0.01).reshape(-1, 1)

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='red', s=80, zorder=5, label='Dữ liệu Thực tế')
plt.plot(X_grid, lin_reg.predict(X_grid), color='gray', linestyle='--', label='Bậc 1 (Linear)')
plt.plot(X_grid, poly_reg_2.predict(X_grid), color='green', label='Bậc 2')
plt.plot(X_grid, poly_reg_3.predict(X_grid), color='orange', label='Bậc 3')
plt.plot(X_grid, poly_reg_4.predict(X_grid), color='blue', linewidth=2.5, label='Bậc 4 (R²=0.997)')
plt.scatter([6.5], [poly_reg_4.predict(test_level)[0]], color='purple', s=160, marker='*', zorder=6, label='Dự đoán Level 6.5')

plt.title('Position Level vs Salary (Polynomial Regression)', fontweight='bold')
plt.xlabel('Position Level')
plt.ylabel('Salary ($)')
plt.legend()
plt.show()
````

📌 **Expected Output:**
- Biểu đồ đường cong bậc 4 màu xanh uốn lượn chính xác qua từng điểm dữ liệu và dự đoán mức lương Level 6.5 là $\approx \$158,862$.
