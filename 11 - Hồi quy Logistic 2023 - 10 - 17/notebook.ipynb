# 📘 Bài Giảng Thực Hành: Hồi Quy Logistic (Logistic Regression)
**Mục tiêu bài giảng:**
1. Hiểu hàm Sigmoid: $P(y=1|\mathbf{x}) = \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}}$.
2. Tối ưu hàm mất mát Log-Loss bằng solver Quasi-Newton (`lbfgs`).
3. Trực quan hóa ranh giới phân lớp tuyến tính (Linear Decision Boundary).

---
### 📊 Tập dữ liệu: `Social_Network_Ads.csv`


## 1. Nạp Dữ Liệu & Huấn Luyện Logistic Regression


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, [2, 3]].values
y = dataset.iloc[:, 4].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)

classifier = LogisticRegression(solver='lbfgs', random_state=0).fit(X_train_scaled, y_train)
print(f'Hệ số trọng số w: {classifier.coef_[0]}')
print(f'Hệ số tự do b:    {classifier.intercept_[0]:.4f}')
print(f'Accuracy:         {accuracy_score(y_test, classifier.predict(X_test_scaled))*100:.2f}%')
```

📌 **Expected Output:**
`Accuracy: 89.00%`


## 2. Trực quan hóa Hàm Sigmoid và Decision Boundary Tuyến Tính


```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 1. Đồ thị Sigmoid
z = np.linspace(-6, 6, 200)
axes[0].plot(z, 1 / (1 + np.exp(-z)), color='purple', linewidth=2.5)
axes[0].axhline(0.5, color='gray', linestyle='--')
axes[0].set_title('Hàm Sigmoid P(y=1|x)', fontweight='bold')
axes[0].set_xlabel('Log-Odds (z)')

# 2. Decision Boundary trên Test Set
X1, X2 = np.meshgrid(np.arange(X_test_scaled[:,0].min()-1, X_test_scaled[:,0].max()+1, 0.02), np.arange(X_test_scaled[:,1].min()-1, X_test_scaled[:,1].max()+1, 0.02))
Z = classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape)
axes[1].contourf(X1, X2, Z, alpha=0.3, cmap=ListedColormap(('salmon', 'lightgreen')))
for i, j in enumerate(np.unique(y_test)):
    axes[1].scatter(X_test_scaled[y_test==j, 0], X_test_scaled[y_test==j, 1], color=ListedColormap(('red', 'green'))(i), label=f'Class {j}')
axes[1].set_title('Logistic Regression (Linear Decision Boundary)', fontweight='bold')
axes[1].legend()
plt.tight_layout()
plt.show()
```

📌 **Expected Output:**
Đồ thị đường cong Sigmoid chữ S và ranh giới phân cách thẳng phân định 2 nhóm dữ liệu.
