# 📘 Bài Giảng Thực Hành: Support Vector Machine Tuyến Tính (Linear SVM)
**Mục tiêu bài giảng:**
1. Hiểu nguyên lý Siêu phẳng Phân cách Tối đa Lề (Maximum Margin Hyperplane): $\mathbf{w}^T \mathbf{x} + b = 0$.
2. Xác định các Vector Hỗ trợ (Support Vectors) và 2 siêu phẳng lề: $\mathbf{w}^T \mathbf{x} + b = \pm 1$.
3. Trực quan hóa chi tiết siêu phẳng, lề dung sai và vòng tròn Support Vectors.

---
### 📊 Tập dữ liệu: `Social_Network_Ads.csv`


## 1. Nạp Dữ Liệu & Huấn Luyện Linear SVM


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, [2, 3]].values
y = dataset.iloc[:, 4].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)

classifier = SVC(kernel='linear', C=1.0, random_state=0).fit(X_train_scaled, y_train)
print('Trọng số w:', classifier.coef_[0])
print('Hệ số tự do b:', classifier.intercept_[0])
print(f'Số lượng Support Vectors: {len(classifier.support_vectors_)}')
print(f'Accuracy trên Test Set:   {accuracy_score(y_test, classifier.predict(X_test_scaled))*100:.2f}%')
```

📌 **Expected Output:**
`Accuracy trên Test Set: 90.00%`
`Số lượng Support Vectors: ~66`


## 2. Trực quan hóa Siêu Phẳng, Lề và Support Vectors


```python
fig, ax = plt.subplots(figsize=(10, 6))
xlim = (X_train_scaled[:, 0].min() - 0.5, X_train_scaled[:, 0].max() + 0.5)
ylim = (X_train_scaled[:, 1].min() - 0.5, X_train_scaled[:, 1].max() + 0.5)
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 300), np.linspace(ylim[0], ylim[1], 300))
xy = np.vstack([xx.ravel(), yy.ravel()]).T
Z = classifier.decision_function(xy).reshape(xx.shape)

# Vùng quyết định và 3 đường siêu phẳng
ax.contourf(xx, yy, (Z > 0).astype(int), alpha=0.3, cmap=ListedColormap(('salmon', 'lightgreen')))
ax.contour(xx, yy, Z, colors='k', levels=[-1, 0, 1], linestyles=['--', '-', '--'], linewidths=[1.5, 2.5, 1.5])

# Điểm dữ liệu
for i, j in enumerate(np.unique(y_train)):
    ax.scatter(X_train_scaled[y_train==j, 0], X_train_scaled[y_train==j, 1], color=ListedColormap(('red', 'green'))(i), label=f'Class {j}')

# Khoanh tròn Support Vectors
sv = classifier.support_vectors_
ax.scatter(sv[:, 0], sv[:, 1], s=120, facecolors='none', edgecolors='blue', label=f'Support Vectors ({len(sv)})')
ax.set_title('Linear SVM (Hyperplane, Margins & Support Vectors)', fontweight='bold')
ax.legend()
plt.show()
```

📌 **Expected Output:**
Biểu đồ trực quan đường thẳng phân cách $w^Tx+b=0$ (nét liền), 2 đường lề $\pm 1$ (nét đứt) và các vòng tròn màu xanh lam khoanh đúng các Support Vectors.
